**Challenge 1: Predict the Output**

Objective:

To predict and investigate the measurement outcomes of single-qubit quantum circuits using Qiskit.

Predictions Before Execution

Circuit A: One X gate flips the initial state from 0 to 1. Expected result: 100 counts of 1.

Circuit B: Two X gates flip the qubit twice, returning it to state 0. Expected result: 100 counts of 0.

Circuit C: One H gate creates an equal superposition of 0 and 1. Expected result: approximately 50 counts of each outcome. The exact counts may vary.

```python
from qiskit import QuantumCircuit

# Create a circuit with 1 qubit and 1 classical bit
qc_a = QuantumCircuit(1, 1)

# Apply one X gate
qc_a.x(0)

# Measure qubit 0 and store the result in classical bit 0
qc_a.measure(0, 0)

# Display the circuit
qc_a.draw("mpl")
```

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Create a local quantum simulator
simulator = AerSimulator()

# Execute Circuit A for 100 shots
result_a = simulator.run(qc_a, shots=100).result()

# Get the measurement counts
counts_a = result_a.get_counts()

# Display the counts
print(counts_a)

# Display the histogram
plot_histogram(counts_a)
```

### Circuit A: One X Gate

**Prediction:** The qubit starts in state `|0⟩`. Applying one X gate flips it to `|1⟩`. Therefore, I predicted that all 100 shots would produce outcome `1`.

**Actual result:** `{'1': 100}`

**Observation:** The histogram shows 100 counts for outcome `1` and zero counts for outcome `0`.

**Conclusion:** The actual result matches my prediction. The X gate flips the qubit from `0` to `1`.
