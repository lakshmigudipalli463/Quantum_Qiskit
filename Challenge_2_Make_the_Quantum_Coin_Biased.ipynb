# Challenge 2: Make the Quantum Coin Biased

## Objective
To create a single-qubit quantum circuit in which the probabilities of measuring 0 and 1 are different.

## Concept
A Hadamard gate creates an equal superposition of 0 and 1, giving each outcome a 50% probability. In this challenge, I will use quantum gates to change the measurement probabilities and observe the results using 100 shots.

```python
from qiskit import QuantumCircuit

# Create a circuit with exactly 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply an Ry gate with a rotation of 60 degrees
import numpy as np
qc.ry(np.pi / 3, 0)

# Measure the qubit
qc.measure(0, 0)

# Display the circuit
qc.draw("mpl")
```

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Create the simulator
simulator = AerSimulator()

# Run the circuit for 100 shots
result = simulator.run(qc, shots=100).result()

# Get and print the measurement counts
counts = result.get_counts()
print(counts)

# Plot the histogram
plot_histogram(counts)
```

## Circuit 1: Bias Towards 0

### Prediction
I used an Ry gate with a rotation angle of 60°. I predicted that approximately 75% of the measurements would produce 0 and 25% would produce 1.

### Actual Result
- Outcome 0: 77 counts (77%)
- Outcome 1: 23 counts (23%)

### Observation
The histogram shows that outcome 0 occurred more frequently than outcome 1. The result is close to my prediction, although the exact counts can vary between runs.

### Conclusion
The Ry gate changes the qubit's state and creates unequal measurement probabilities. With a 60° rotation, outcome 0 is more likely than outcome 1.

```python
# Create a second circuit with 1 qubit and 1 classical bit
qc2 = QuantumCircuit(1, 1)

# Rotate the qubit by 120 degrees
qc2.ry(2 * np.pi / 3, 0)

# Measure the qubit
qc2.measure(0, 0)

# Display the circuit
qc2.draw("mpl")
```

```python
# Run Circuit 2 for 100 shots
result2 = simulator.run(qc2, shots=100).result()

# Get and print the measurement counts
counts2 = result2.get_counts()
print(counts2)

# Plot the histogram
plot_histogram(counts2)
```

## Circuit 2: Bias Towards 1

### Prediction
I used an Ry gate with a rotation angle of 120°. I predicted that approximately 25% of the measurements would produce 0 and 75% would produce 1.

### Actual Result
- Outcome 0: 32 counts (32%)
- Outcome 1: 68 counts (68%)

### Observation
The histogram shows that outcome 1 occurred more frequently than outcome 0. The actual counts differ slightly from the predicted probabilities because quantum measurement outcomes are probabilistic.

### Conclusion
Increasing the Ry rotation angle from 60° to 120° changes the measurement probabilities. In this circuit, outcome 1 becomes more likely than outcome 0.

Both circuits use one qubit and one Ry gate before measurement. 
Changing the rotation angle changes the measurement probabilities.
The 60° rotation makes outcome 0 more likely, while the 120° rotation makes outcome 1 more likely.

## Challenge Question

**How can quantum gates change the probability of obtaining 0 or 1 when a qubit is measured?**

Quantum gates change the quantum state of a qubit by changing its probability amplitudes. When the qubit is measured, these amplitudes determine the probabilities of obtaining 0 or 1. For example, the Ry gate changes these probabilities depending on its rotation angle. A 60° rotation makes 0 more likely, while a 120° rotation makes 1 more likely.
