
# Challenge 4: Quantum Correlation Switch

## Objective
To design two quantum circuits that produce correlated and opposite measurement outcomes.

## Concept
A Bell-state circuit can produce correlated outcomes, where both qubits give the same result. By adding an X gate to one qubit, we can switch to opposite outcomes, where the two qubits give different results.

## Tools Used
- Python
- Qiskit
- Qiskit Aer Simulator
- Matplotlib histogram



## Predictions Before Execution

### Part A – Correlated Outcomes
I predict that the outcomes 00 and 11 will dominate. Both qubits should produce the same measurement result because the circuit creates a Bell state.

### Part B – Opposite Outcomes
I predict that the outcomes 01 and 10 will dominate. Adding an X gate to qubit 0 should flip its result, making the two qubits produce opposite measurement results.

### Will All Four Outcomes Appear?
I predict that only two outcomes will appear in each circuit. In an ideal simulation, Part A produces 00 and 11, while Part B produces 01 and 10. The exact counts may vary slightly across 100 shots.


```python

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Create a circuit with exactly 2 qubits and 2 classical bits
qc_a = QuantumCircuit(2, 2)

# Create correlated outcomes using a Bell-state circuit
qc_a.h(0)
qc_a.cx(0, 1)

# Measure both qubits
qc_a.measure([0, 1], [0, 1])

# Display the circuit
display(qc_a.draw("mpl"))
```

```python

# Create the simulator
simulator = AerSimulator()

# Run Part A for 100 shots
compiled_a = transpile(qc_a, simulator)
result_a = simulator.run(compiled_a, shots=100).result()

# Display measurement counts
counts_a = result_a.get_counts()
print("Part A - Correlated Outcomes:", counts_a)

# Display the histogram
plot_histogram(counts_a)
```

```python

# Create Part B circuit
qc_b = QuantumCircuit(2, 2)

# Prepare the Bell state
qc_b.h(0)
qc_b.cx(0, 1)

# Add one X gate to switch the correlation
qc_b.x(0)

# Measure both qubits
qc_b.measure([0, 1], [0, 1])

# Display the modified circuit
display(qc_b.draw("mpl"))
```

```python

# Run Part B for 100 shots
compiled_b = transpile(qc_b, simulator)
result_b = simulator.run(compiled_b, shots=100).result()

# Display measurement counts
counts_b = result_b.get_counts()
print("Part B - Opposite Outcomes:", counts_b)

# Display the histogram
plot_histogram(counts_b)
```


## Part C – Comparison of Both Experiments


### Observations

**Part A – Correlated Outcomes**
The outcomes 00 and 11 were dominant, appearing 51 and 49 times, respectively. Both qubits produced the same measurement result.

**Part B – Opposite Outcomes**
The outcomes 01 and 10 were dominant, appearing 47 and 53 times, respectively. The two qubits produced opposite measurement results.

### Explanation

The Part A circuit uses a Hadamard gate followed by a CNOT gate to prepare a Bell state. This creates correlated outcomes, so the qubits produce the same result.

In Part B, an X gate is applied to qubit 0 after preparing the Bell state. The X gate flips that qubit's measurement result, changing the possible outcomes from 00 and 11 to 01 and 10.

Thus, adding only one quantum gate switches the measurement relationship from correlated to opposite. Both circuits use exactly two qubits and were executed for 100 shots each.

### Conclusion

The experiment demonstrates that quantum measurement correlations can be changed by modifying a quantum circuit. The original circuit produced matching outcomes, while the modified circuit produced opposite outcomes. The results matched the predictions.
